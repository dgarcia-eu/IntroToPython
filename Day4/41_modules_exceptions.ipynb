You can rename a module with **as** and call its functions or variables with the alias you used to name it:

```python
import statistics as s     #  we name statistics -> s
s.mean([1,2,3])
```

```python
from statistics import mean # we only import mean function from statistics
mean([1,2,3])               # we don't need to call statistics anymore. BUT this is not recommendable to avoide name crush.
```

You can make your own module with functions you want to reuse across scripts. [This example from the official python documentation](https://docs.python.org/3/tutorial/modules.html) shows how to do it:

- A module is just a text file ending in **.py** that sits next to your notebook. The cell below writes one called **fibo.py**. The `%%writefile` in the first line is a Jupyter command that saves the rest of the cell to that file instead of running it -- so `fib` and `fib2` are *not* defined in this notebook, only inside the file.
- If you prefer, you can create fibo.py yourself in a text editor and paste the same code into it.

```python
%%writefile fibo.py
# Fibonacci numbers module

def fib(n):
    """Write Fibonacci series up to n."""
    a, b = 0, 1
    while a < n:
        print(a, end=' ')
        a, b = b, a+b
    print()

def fib2(n):
    """Return Fibonacci series up to n."""
    result = []
    a, b = 0, 1
    while a < n:
        result.append(a)
        a, b = b, a+b
    return result
```

Now the file exists next to the notebook, so you can **import fibo** and use the two functions it defines:

```python
import fibo

fibo.fib(1000)

print(fibo.fib2(100))
```

<font size="+3">Handling exceptions</font>

Sometimes you don't want a program to stop when an error happens, so you can run some code for that case. Python has exceptions as a way to cope with unexpected code execution: https://docs.python.org/3/library/exceptions.html#BaseException 
For example, each error in the code raises an exception, in this case NameError:

```python
# Python raises exceptions when it encounter errors. For example..

Print('Hello!') # -> NameError
```

### Catching exceptions

You can write you code to catch exceptions with **try, except** cases.

In the following example, we have a list called short_list containing both integers and a string. The goal is to sum all integers in the list. However, adding a string to an integer will raise an exception, so we use try and except to handle this:

```python
short_list = [13, 45, 2, 'num', 5, 100]

sum = 0
for i in short_list:
    sum += i
```

We can use **try** and **except** to catch the exception of the error and prevent the code from stopping there, doing nothing with *pass*:

```python
# try and except
short_list = [13, 45, 2, 'two', 5, 100]

sum = 0
for i in short_list:
    try:
        sum += i                    # Try this statemtent 
    except:                         # But if there is an exception
        pass                        # Do nothing
print(sum)                   
```

Or printing the item that caused the exception:

```python
short_list = [13, 45, 2, 'two', 5, 100]

sum = 0
for i in short_list:
    try:
        sum += i                    # Try this statement
    except Exception as e:          # Catch any exception that occurs and stores the error message in e
        print("{}: {}".format(i, e))  # Print the item that caused an error and the error message

print("Total sum:", sum)

```

You can catch different types of exceptions. You can also add an **else** block, which runs only if no exception was raised, and a **finally** block, which runs in every case (exception or not):

```python
short_list = [13, 45, 2, 'two', 5, 100]

sum = 0
for i in short_list:
    try:
        sum += i               # Try this statement
    except TypeError as t:     # If there is a TypeError
        print("TypeError {}: {}".format(i, t))
    except Exception as e:          # Catch any exception that occurs and stores the error message in e
        print("{}: {}".format(i, e))  # Print the item that caused an error and the error message
    else:
        print("No exceptions!")    # else: if there was no exception, execute it.
    finally:
        print("finished\n")        # Execute it regardless of exceptions.
print("Total sum:", sum)
```

### 2. `TypeError` — right name, wrong kind of thing

You asked Python to combine two things that do not go together. Nearly always a number where a string was expected or the other way round.

**Fix:** print the values and check what they actually are. `type(x)` tells you.

```python
age = 24
print("I am " + age)
```

### 3. `IndexError` / `KeyError` — that position or key is not there

`IndexError` for lists, `KeyError` for dictionaries. Usually an off-by-one, or a key you assumed was present.

**Fix:** print the thing and its `len()` before you index it. For dictionaries, `.get(key, default)` returns a fallback instead of raising.

```python
names = ['ada', 'grace', 'alan']
print(names[3])
```

### 4. `AttributeError` — this type cannot do that

You called a method that exists for some other type. `.append()` on a tuple, `.lower()` on a list, `.split()` on a number.

**Fix:** check what the variable actually is. It is usually one level off — a list where you meant one item of it.

```python
words = ['the', 'old', 'man']
print(words.lower())
```

### 5. `IndentationError` / `SyntaxError` — Python cannot even read it

These are different from the others: nothing ran at all, so the reported line is where Python gave up, and **the real mistake is often the line above** — a missing colon, an unclosed bracket or quote.

**Fix:** look at the line before the one it points at.

### The worst kind: no error at all

The code below runs perfectly and gives the wrong answer. There is no traceback to read, so the only way to find it is to check whether the result is *plausible*.

**Predict what it should print, then run it.**

```python
words = ['the', 'old', 'man', 'and', 'the', 'sea']

for word in words:
    total = 0
    total = total + len(word)

print("total characters:", total)
```

It printed `3`. The six words have 19 characters between them.

`total = 0` is inside the loop, so it resets on every pass and you get the length of the last word. Nothing is wrong syntactically; the code just does not mean what it looks like it means.

**This is why you check that an answer is plausible before you believe it.** Ask yourself: six short words — should that be about 3, or about 20?

The general technique: when a result is wrong but nothing errored, print things *inside* the loop and watch them change.

```python
words = ['the', 'old', 'man', 'and', 'the', 'sea']

for word in words:
    total = 0
    total = total + len(word)
    print("  after", word, "total is", total)   # <- watch it reset

print("total characters:", total)
```
