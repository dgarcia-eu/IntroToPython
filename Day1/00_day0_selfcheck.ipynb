## Part 3 — where are you starting from?

Not a test. Nothing is recorded. Just read each block and ask yourself: *could I have written that?*

---

**Level 1.** Storing something and printing it.

```python
message = 'Python is a language'
print(message)
print(message.upper())
```

**Level 2.** Doing something to every item of a list.

```python
words = ['the', 'old', 'man', 'and', 'the', 'sea']
for word in words:
    print(word, len(word))
```

**Level 3.** Building up an answer while looping.

```python
words = ['the', 'old', 'man', 'and', 'the', 'sea']
total = 0
for word in words:
    total = total + len(word)
print('total characters:', total)
```

**Level 4.** The same thing, written the short way.

```python
words = ['the', 'old', 'man', 'and', 'the', 'sea']
print(sum([len(w) for w in words]))
```
