# A Python file is just a text file

You already made one on Day 4: `fibo.py`, with `%%writefile`. There is nothing special about it — a `.py` file is a text file containing Python, and running it executes every line from top to bottom, once, in order.

That is the whole difference from a notebook. No cells, no out-of-order execution, no leftover variables from something you ran ten minutes ago.

````python
%%writefile word_count.py
"""Count the words in a piece of text. Our first real script."""

text = "the old man and the sea"
words = text.split()

counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1

print("words:", len(words))
print("counts:", counts)
````

# Running it from a terminal

In Jupyter you can run a shell command from a notebook by putting `!` in front of it. That is what the next cell does — but the point is that the command itself is what you would type in a terminal, without the `!`:

```
python word_count.py
```

````python
!python word_count.py
````

Open a terminal and try it for real. **Anaconda Prompt** on Windows, **Terminal** on macOS and Linux. You will need three commands:

| command | what it does |
|---|---|
| `cd some_folder` | go into a folder |
| `cd ..` | go back up one |
| `ls` (macOS/Linux) or `dir` (Windows) | list what is here |

Navigate to the folder holding `word_count.py`, then run `python word_count.py`. This is how nearly all Python outside a notebook gets run, including everything on a server, and it is what ICSS will expect from next week.

# Installing packages

`pandas` and `matplotlib` came with Anaconda. Most packages do not, and you install them yourself. Two commands do this — use `conda` if you installed Anaconda, `pip` otherwise:

```
conda install requests        # the Anaconda way
pip install requests          # works everywhere
```

You can also run them from a notebook with `!`, though a terminal is better. Let's check what is already installed:

````python
!pip list | head -20
````

**Where does an installed package go?** Into the Python installation you are currently using — which is the thing the next section is about.

# Virtual environments

Here is the problem they solve. Suppose one project needs pandas 1.5 and another needs pandas 3.0. There is one Python installation on your laptop, so installing one version breaks the other project. This bites everyone eventually.

A **virtual environment** is a separate, self-contained set of packages. Each project gets its own, and they cannot interfere with each other.

With conda:

```
conda create -n icss python=3.13     # make one called 'icss'
conda activate icss                  # start using it
conda install pandas matplotlib      # installs into 'icss' only
conda deactivate                     # go back to normal
```

When an environment is active its name appears in your prompt, like `(icss) $`. That is how you know which one you are in — and *which environment am I in?* is the answer to a surprising share of "but I installed it!" problems.

This repository has an `environment.yml` describing exactly what the course needs, so you could recreate it anywhere with:

```
conda env create -f environment.yml
conda activate introtopython
```
